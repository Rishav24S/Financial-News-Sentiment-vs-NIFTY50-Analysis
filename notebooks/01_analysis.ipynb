```python
def sentiment_score(text):
    text = str(text).lower()
    score = 0
    for word in positive_words:
        score += text.count(word)
    for word in negative_words:
        score -= text.count(word)
    return score
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import TimeSeriesSplit, GridSearchCV
import xgboost as xgb
```

```python
ml_df = pd.read_csv("../data/final_project_dataset.csv", parse_dates=['date'])
ml_df = ml_df.sort_values('date').reset_index(drop=True)

# Shift target to predict close(t+1) > close(t)
ml_df['next_day_target'] = ml_df['target'].shift(-1)
ml_df = ml_df.dropna(subset=['next_day_target']).copy()
ml_df['next_day_target'] = ml_df['next_day_target'].astype(int)
```

```python
ml_df['sent_ema3'] = ml_df['sentiment_score'].ewm(span=3).mean()
ml_df['sent_ema7'] = ml_df['sentiment_score'].ewm(span=7).mean()

# Lagged price dynamics & volatility indicators
ml_df['movement_lag1'] = ml_df['movement'].shift(1)
ml_df['movement_lag2'] = ml_df['movement'].shift(2)
ml_df['volume_change'] = ml_df['shares_traded'].pct_change()
ml_df['high_low_spread'] = (ml_df['high'] - ml_df['low']) / ml_df['open'] * 100

ml_df = ml_df.dropna().reset_index(drop=True)

feature_cols = [
    'sentiment_score', 'sent_ema3', 'sent_ema7',
    'movement_lag1', 'movement_lag2', 'volume_change', 'high_low_spread'
]

X = ml_df[feature_cols]
y = ml_df['next_day_target']
```

```python
split_idx = int(len(ml_df) * 0.8)
X_train, X_test = X.iloc[:split_idx], X.iloc[split_idx:]
y_train, y_test = y.iloc[:split_idx], y.iloc[split_idx:]

majority_baseline = (y_test == y_test.mode()[0]).mean()
print(f"Majority-Class Baseline (Test Period): {majority_baseline * 100:.2f}%\n")
```

```python
# Logistic Regression
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

lr = LogisticRegression(max_iter=1000, random_state=42)
lr.fit(X_train_scaled, y_train)
lr_acc = accuracy_score(y_test, lr.predict(X_test_scaled))

# Tuned Random Forest using TimeSeriesSplit
tscv = TimeSeriesSplit(n_splits=5)
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [3, 5],
    'min_samples_leaf': [5, 10]
}

rf_grid = GridSearchCV(
    RandomForestClassifier(random_state=42), 
    param_grid, cv=tscv, scoring='accuracy', n_jobs=-1
)
rf_grid.fit(X_train, y_train)
best_rf = rf_grid.best_estimator_
rf_acc = accuracy_score(y_test, best_rf.predict(X_test))

print(f"Logistic Regression Test Accuracy: {lr_acc * 100:.2f}%")
print(f"Tuned Random Forest Test Accuracy: {rf_acc * 100:.2f}%\n")
print(classification_report(y_test, best_rf.predict(X_test)))
```

```python
cm = confusion_matrix(y_test, best_rf.predict(X_test))
plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Purples', xticklabels=['Down', 'Up'], yticklabels=['Down', 'Up'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Tuned Random Forest Confusion Matrix')
plt.tight_layout()
plt.savefig('../plots/confusion_matrix.png', dpi=300)
plt.show()
```

```python
test_df = ml_df.iloc[split_idx:].copy().reset_index(drop=True)
test_df['pred'] = best_rf.predict(X_test)
test_df['next_day_return_pct'] = (test_df['close'].shift(-1) - test_df['close']) / test_df['close'] * 100
test_df['strategy_return'] = np.where(test_df['pred'] == 1, test_df['next_day_return_pct'], 0)

test_df['cum_strategy'] = (1 + test_df['strategy_return'].fillna(0) / 100).cumprod()
test_df['cum_market'] = (1 + test_df['next_day_return_pct'].fillna(0) / 100).cumprod()

plt.figure(figsize=(8, 4))
plt.plot(test_df['date'], test_df['cum_strategy'], label='Model Strategy', color='purple')
plt.plot(test_df['date'], test_df['cum_market'], label='Buy & Hold NIFTY50', color='grey', linestyle='--')
plt.legend()
plt.title('Backtest: Strategy vs Buy-and-Hold (Test Period)')
plt.ylabel('Growth of ₹1')
plt.xticks(rotation=45)
plt.gcf().autofmt_xdate()
plt.tight_layout()
plt.savefig('../plots/backtest.png', dpi=300)
plt.show()
```
